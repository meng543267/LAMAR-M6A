load_from_disk(dataset)

EsmForSequenceClassification

新版支持 `full / mature`、任意细胞系和长度，并把结果保存到 `outputs/`。


## Step 0. 参数设置


````python
from pathlib import Path
import os
import sys

PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "lamar_m6a").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

print(PROJECT_ROOT)
````

````python
MODE = "full"  # full / mature
CELL = "hek293"
LENGTH = 801

DATASET_DIR = Path("dataset")
DATA_PATH = DATASET_DIR / MODE / f"m6a_{MODE}_{CELL}_{LENGTH}"
OUTPUTS_DIR = Path("outputs")
OUTPUT_DIR = OUTPUTS_DIR / "checkpoints" / MODE / f"{CELL}-{LENGTH}-bs16-acc4"

TOKENIZER_PATH = Path("tokenizer/single_nucleotide")
MODEL_CONFIG = Path("configs/config_150M.json")
PRETRAIN_STATE_PATH = Path("pretrain_model/model.safetensors")
MODEL_CLASS = "scripts.sequence_classification_patch:EsmForSequenceClassification"
FLASH_ATTENTION_CLASS = "flash_attn_patch:EsmSelfAttentionAddFlashAttnPatch"

MODEL_MAX_LENGTH = LENGTH + 2
NUM_LABELS = 2
TOKEN_DROPOUT = False
POSITIONAL_EMBEDDING_TYPE = "rotary"
HIDDEN_SIZE = 768
INTERMEDIATE_SIZE = 3072
NUM_ATTENTION_HEADS = 12
NUM_HIDDEN_LAYERS = 12
HEAD_TYPE = "Linear"
FREEZE = False
KERNEL_SIZES = [2, 3, 5]
OCS = 32
FLASH_ATTENTION = False

BATCH_SIZE = 16
EVAL_BATCH_SIZE = 16
GRADIENT_ACCUMULATION_STEPS = 4
LEARNING_RATE = 1e-5
WARMUP_RATIO = 0.05
EPOCHS = 2
MAX_GRAD_NORM = 1.0
EVAL_STEPS = 100
SAVE_STEPS = 100
LOGGING_STEPS = 100
FP16 = False

TRAIN_SPLIT = "train"
EVAL_SPLIT = "test"

DATA_PATH, OUTPUT_DIR
````

## Step 1. 导入函数和依赖


````python
from lamar_m6a.finetune.finetune_m6a import (
    compute_metrics,
    import_object,
    load_pretrained_weights,
    make_training_arguments,
)

import json
from datasets import load_from_disk
from transformers import AutoConfig, AutoTokenizer, DataCollatorWithPadding, Trainer
````

## Step 2. 读取 tokenized dataset


````python
data = load_from_disk(DATA_PATH)
print(data)
for split in data.keys():
    print(split, data[split].column_names, len(data[split]))
````

## Step 3. 加载 tokenizer 和 model config


````python
tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_PATH, model_max_length=MODEL_MAX_LENGTH)

data_collator = DataCollatorWithPadding(tokenizer=tokenizer, padding=True)

config = AutoConfig.from_pretrained(
    MODEL_CONFIG,
    vocab_size=len(tokenizer),
    pad_token_id=tokenizer.pad_token_id,
    mask_token_id=tokenizer.mask_token_id,
    num_labels=NUM_LABELS,
    token_dropout=TOKEN_DROPOUT,
    positional_embedding_type=POSITIONAL_EMBEDDING_TYPE,
    hidden_size=HIDDEN_SIZE,
    intermediate_size=INTERMEDIATE_SIZE,
    num_attention_heads=NUM_ATTENTION_HEADS,
    num_hidden_layers=NUM_HIDDEN_LAYERS,
)
config
````

## Step 4. 构建模型并加载预训练参数


````python
model_cls = import_object(MODEL_CLASS)
model = model_cls(
    config,
    head_type=HEAD_TYPE,
    freeze=FREEZE,
    kernel_sizes=KERNEL_SIZES,
    ocs=OCS,
)

if FLASH_ATTENTION:
    flash_cls = import_object(FLASH_ATTENTION_CLASS)
    for i in range(config.num_hidden_layers):
        model.esm.encoder.layer[i].attention.self = flash_cls(
            config,
            position_embedding_type=POSITIONAL_EMBEDDING_TYPE,
        )

load_pretrained_weights(model, PRETRAIN_STATE_PATH)
model
````

## Step 5. 设置 TrainingArguments


````python
class Args:
    dataloader_drop_last = True
    batch_size = BATCH_SIZE
    eval_batch_size = EVAL_BATCH_SIZE
    learning_rate = LEARNING_RATE
    weight_decay = 0.01
    adam_beta1 = 0.9
    adam_beta2 = 0.98
    adam_epsilon = 1e-8
    warmup_ratio = WARMUP_RATIO
    epochs = EPOCHS
    max_grad_norm = MAX_GRAD_NORM
    gradient_accumulation_steps = GRADIENT_ACCUMULATION_STEPS
    eval_steps = EVAL_STEPS
    save_steps = SAVE_STEPS
    logging_steps = LOGGING_STEPS
    fp16 = FP16
    report_to = "none"
    save_total_limit = 3
    load_best_model_at_end = False
    metric_for_best_model = "pr_auc"

train_args = make_training_arguments(Args, OUTPUT_DIR)
train_args
````

## Step 6. Trainer 训练


````python
trainer = Trainer(
    model=model,
    args=train_args,
    train_dataset=data[TRAIN_SPLIT],
    eval_dataset=data[EVAL_SPLIT],
    compute_metrics=compute_metrics,
    data_collator=data_collator,
    tokenizer=tokenizer,
)

trainer.train()
trainer.save_model(OUTPUT_DIR / "final_model")
````

## Step 7. eval 验证并保存 metrics

这里只保存训练过程 eval metrics。当前默认使用 `test` split 作为 fine-tune 阶段的 eval split；后续独立验证/测试交给专门的 evaluation 脚本处理。


````python
metrics_dir = OUTPUTS_DIR / "metrics" / MODE
metrics_dir.mkdir(parents=True, exist_ok=True)

valid_metrics = trainer.evaluate(data[EVAL_SPLIT])
metrics_path = metrics_dir / f"{CELL}_{LENGTH}_{EVAL_SPLIT}.json"
metrics_path.write_text(
    json.dumps(valid_metrics, indent=2, ensure_ascii=False),
    encoding="utf-8",
)
print("valid:", valid_metrics)
print("saved validation metrics:", metrics_path)
````
